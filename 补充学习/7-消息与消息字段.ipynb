## 消息对象的字段

### SystemMessage
系统消息，用于设置模型的行为或提供上下文。

只有一个字段：content

### HumanMessage
用户消息，用于向模型发送用户输入。
**content**：消息内容，字段名可以省略

`HumanMessage("你好啊~")`相当于`HumanMessage(content = "你好啊~")`

**metadata**：元数据字段，可以有很多，自定义

举例：带有元数据字段
```python
HumanMessage(
    content="Hello!",
    name="alice",  # 可选，用户名
    id="msg_123",  # 可选，message的ID
)
```

`name` 和 `id` 都属于元数据字段，当消息类型相同，对消息进行区分。但不是所有模型都支持这一功能，是否支持取决于模型供应商，需要查看官方手册。比如：

[OpenAI的API手册](https://developers.openai.com/api/reference/resources/chat/subresources/completions/methods/create)告诉我们，HumanMessage支持 `name` 作为元数据字段，如下图所示。而 DeepSeek的API官方文档明确支持 `name` 作为元数据，但实测发现模型无法识别。

````python
from langchain_core.messages import SystemMessage,HumanMessage,AIMessage
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os
# 从.env文件中加载环境变量
load_dotenv(override=True)
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
OPENROUTER_BASE_URL = os.getenv("OPENROUTER_BASE_URL")
model = init_chat_model(
    model="gpt-5.4-mini",
    model_provider="openai",
    api_key=OPENROUTER_API_KEY,
    base_url=OPENROUTER_BASE_URL
)

messages = [
    SystemMessage("你是一个信息抽取器。你会收到多条来自不同发言者的 user 消息。每条消息可能带有 name 字段。你的任务是：严格根据每条消息的 name 提取发言者及其观点，并输出 JSON。禁止使用“第一个人/第二个人”这种相对称呼。若某条消息没有 name，则输出 unknown。输出格式：{\"speakers\":[{\"name\":\"...\",\"claim\":\"...\"}]}"),
    HumanMessage(
        content="我认为 1+1=2",
        name="Bob"),
    HumanMessage(
        content="我认为 1+1>2",
        name="Tom"),
    HumanMessage(
        content="请列出谁说了什么，不要判断对错。",
        name="audience"
    )
]
response = model.invoke(messages)
print(response.content)
````

### AIMessage

这些字段，在《invoke的返回值》中讲过

**content**：模型输出的原始内容，字段名可以省略
`AIMessage("你好~")`相当于`AIMessage(content="你好~")`

**response_metadata**：AIMessage特有属性，LLM的响应中附加元数据，根据不同模型会有不同，如可能会包含本次token使用量等信息。

**usage_metadata**：用量信息。

**tool_calls**：AIMessage特有属性，表示工具调用信息。当LLM决定调用工具时，在AIMessage中就会包含这个属性，没有工具调用则为空。结构如下：
```python
AIMessage(
    content="",
    tool_calls=[
    {
        'name': 'get_weather',  # 应调用的工具名
        'args': {'city': '杭州'},  # 调用工具的参数
        'id': 'call_00_gIXYOD1Q1OkEXmdDBqXR1578', # 工具调用的唯一标识ID
        'type': 'tool_call'
    },
    {'name': 'get_news', 
     'args': {}, 
     'id': 'call_01_jD3phD5PEaIZf0mVLhKt0861', 
     'type': 'tool_call'
    }
]
)
```


````python
from langchain_core.messages import SystemMessage,HumanMessage
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os
from rich import print as rprint
# 从.env文件中加载环境变量
load_dotenv(override=True)
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
OPENROUTER_BASE_URL = os.getenv("OPENROUTER_BASE_URL")
model = init_chat_model(
    model="gpt-5.4-mini",
    model_provider="openai",
    api_key=OPENROUTER_API_KEY,
    base_url=OPENROUTER_BASE_URL
)
messages = [
    SystemMessage("你叫小智，是一名助人为乐的助手。"),
    HumanMessage("你好，好久不见，请介绍下你自己。")
]
response = model.invoke(messages)
rprint(response)
````

### ToolMessage
**content**：工具输出内容

**name**：工具名称

**tool_call_id**：工具调用唯一ID，ToolMessage必须紧邻匹配的AIMessage，和前者tool_calls中的id一致。

```python
ToolMessage(
    content="<工具输出>",
    name="get_weather",
    tool_call_id="call_00_nUD2NC9QRN5Cg1GaoIkBJQ4s"  # 一定要和AI消息中的调用ID匹配
)
```

````python
### JSON传递原始工具消息，用于结合业务内容自定义工具消息

from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os
# 从.env文件中加载环境变量
load_dotenv(override=True)
CLOSEAI_API_KEY = os.getenv("CLOSEAI_API_KEY")
CLOSEAI_BASE_URL = os.getenv("CLOSEAI_BASE_URL")
model = init_chat_model(
model="gpt-5.4-mini",
model_provider="openai",
api_key=CLOSEAI_API_KEY,
base_url=CLOSEAI_BASE_URL
)
def get_weather(city: str) -> str:
    return city + "不错哦~"
# 模拟模型绑定工具
model_with_tools = model.bind_tools([get_weather])
ai_message = {
    "role": "assistant",
    "content": "",
    "tool_calls": [
        {
        "name": "get_weather",
        "args": {"location": "北京"},
        "id": "call_00_nUD2NC9QRN5Cg1GaoIkBJQ4s"
    }]
}
tool_message = {
    "role": "tool",
    "content": "今天北京天气晴朗，万里无云~",
    "tool_call_id": "call_00_nUD2NC9QRN5Cg1GaoIkBJQ4s"
}
messages = [
    {"role": "user", "content": "北京天气如何"},
    ai_message,
    tool_message
]
response = model.invoke(messages)
print(response)
````

````python
### 使用@tool装饰工具函数

import os

from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain.tools import tool
from rich import print as rprint

load_dotenv(override=True)
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
OPENROUTER_BASE_URL = os.getenv("OPENROUTER_BASE_URL")

model = init_chat_model(
    model="gpt-5.4-mini",
    model_provider="openai",
    api_key=OPENROUTER_API_KEY,
    base_url=OPENROUTER_BASE_URL,
)


@tool
def get_weather(city: str) -> str:
    """查询指定城市的天气。"""
    return f"{city} 今天天气晴朗，适合出门。"


# 1. bind_tools 会把 Python 函数转换成模型可理解的工具 schema。
# 注意：绑定工具后，模型只会在 AIMessage.tool_calls 中提出“我要调用什么工具”。
# 如果没有使用 Agent，工具不会自动执行，需要我们手动执行并把结果传回模型。
model_with_tools = model.bind_tools([get_weather])

# 为了稳定演示 ToolMessage 链路，第一次调用用 tool_choice 强制模型调用 get_weather。
model_with_tools_forced = model.bind_tools([get_weather], tool_choice="get_weather")

messages = [{"role": "user", "content": "北京今天的天气怎么样？"}]

# 2. 第一次调用：让模型决定是否需要调用工具。
ai_message = model_with_tools_forced.invoke(messages)
messages.append(ai_message)

rprint(ai_message.tool_calls)

# 3. 手动执行模型请求的工具调用。
# get_weather.invoke(tool_call) 会返回 ToolMessage，
# 其中 tool_call_id 会自动和 AIMessage.tool_calls 中的 id 对齐。
for tool_call in ai_message.tool_calls:
    if tool_call["name"] == "get_weather":
        # 当前版本的 Tool.invoke() 识别标准 ToolCall 结构时需要 type 字段。
        # 部分 provider 返回的 tool_call 里已经带有 type；这里补一下，保证示例稳定。
        tool_call = {**tool_call, "type": "tool_call"}
        tool_message = get_weather.invoke(tool_call)
        messages.append(tool_message)

# 4. 第二次调用：把 ToolMessage 传回模型，让模型基于工具结果生成自然语言回答。
final_response = model_with_tools.invoke(messages)
rprint(final_response)
````
