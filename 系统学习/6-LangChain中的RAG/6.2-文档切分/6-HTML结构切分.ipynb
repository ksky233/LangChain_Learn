# HTML 结构切分

HTML 中的 `h1`、`h2`、`h3` 和 section 标签提供了天然结构。LangChain 当前提供多种 HTML Splitter：

- `HTMLHeaderTextSplitter`：根据标题标签切分并保留标题 metadata。
- `HTMLSectionSplitter`：根据 section 结构切分。
- `HTMLSemanticPreservingSplitter`：在长度约束下尽量保留表格、列表等元素。

本例先演示最容易理解的 Header 切分。

```python
from pathlib import Path
from langchain_text_splitters import HTMLHeaderTextSplitter

path = Path.cwd() / "系统学习" / "6-LangChain中的RAG" / "asset" / "load" / "07-load.html"
html = path.read_text(encoding="utf-8")

splitter = HTMLHeaderTextSplitter(
    headers_to_split_on=[
        ("h1", "chapter"),
        ("h2", "section"),
        ("h3", "subsection"),
    ]
)
chunks = splitter.split_text(html)
```

```python
for chunk in chunks:
    print(chunk.metadata)
    print(chunk.page_content[:250])
    print("-" * 50)
```

标题结构切分依赖 HTML 本身的语义标签质量。大量使用 `div`、缺少标题层级或正文由 JavaScript 动态生成时，应先完成网页正文提取和结构清理，再交给 Splitter。
