# TokenTextSplitter：按 Token 切分

字符数不等于 Token 数。Token 切分使用具体 Tokenizer 控制块长度，适合严格匹配模型输入预算。不同模型的 Tokenizer 可能不同，因此应尽量使用与目标模型一致或接近的编码器。

```python
import tiktoken
from langchain_text_splitters import TokenTextSplitter

text = "LangChain 支持文档加载、切分、嵌入、检索以及 Agent 编排。" * 8
encoding = tiktoken.get_encoding("cl100k_base")

splitter = TokenTextSplitter(
    encoding_name="cl100k_base",
    chunk_size=30,
    chunk_overlap=5,
)
chunks = splitter.split_text(text)

for index, chunk in enumerate(chunks, start=1):
    print(index, "chars=", len(chunk), "tokens=", len(encoding.encode(chunk)))
    print(repr(chunk))
```

## 另一种写法

许多 TextSplitter 支持 `from_tiktoken_encoder()`，它保留原切分器的边界策略，只把长度计算改为 Token 数。纯 `TokenTextSplitter` 更严格，但可能直接切在自然语言边界中间。

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

recursive_token_splitter = RecursiveCharacterTextSplitter.from_tiktoken_encoder(
    encoding_name="cl100k_base",
    chunk_size=30,
    chunk_overlap=5,
)
print(recursive_token_splitter.split_text(text)[:2])
```
