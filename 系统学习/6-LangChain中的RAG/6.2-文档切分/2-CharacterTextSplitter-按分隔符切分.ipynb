# CharacterTextSplitter：按分隔符切分

`CharacterTextSplitter` 先使用指定 `separator` 得到小片段，再把相邻片段合并到接近 `chunk_size`。它并不是机械地每隔 N 个字符截断。

```python
from langchain_text_splitters import CharacterTextSplitter

text = (
    "第一段：LangChain 提供统一的组件接口。\n\n"
    "第二段：文档切分影响后续召回质量 。\n\n"
    "第三段：Chunk 应保留来源和位置信息。"
)

splitter = CharacterTextSplitter(
    separator="\n\n",
    chunk_size=35,
    chunk_overlap=0,
)

for index, chunk in enumerate(splitter.split_text(text), start=1):
    print(index, repr(chunk), len(chunk))
```

## 注意

如果某个分隔片段本身已经超过 `chunk_size`，`CharacterTextSplitter` 不一定能继续细分它，可能产生超长块。需要多级兜底时，应使用 `RecursiveCharacterTextSplitter`。
