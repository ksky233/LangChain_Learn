# RecursiveCharacterTextSplitter：递归切分

这是 LangChain 官方建议多数场景优先尝试的通用切分器。它按分隔符优先级递归寻找边界，默认思路接近：

```text
段落 → 换行 → 空格 → 字符
```

只有较大的结构无法满足长度限制时，才继续尝试更细的分隔符。

````python
from pathlib import Path
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter

path = Path.cwd() / "系统学习" / "6-LangChain中的RAG" / "knowledge.txt"
document = Document(
    page_content=path.read_text(encoding="utf-8"),
    metadata={"source": str(path)},
)

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=80,
    add_start_index=True,
)
chunks = splitter.split_documents([document])
````

````python
print(f"Chunk 数量：{len(chunks)}")
for chunk in chunks[:3]:
    print(chunk.metadata)
    print(repr(chunk.page_content[:150]), "...")
````

`chunk_overlap` 表示相邻块尝试重复保留的内容长度，并不保证每一对 Chunk 都精确重叠相同字符数。实际重叠还受分隔符和片段合并结果影响。
