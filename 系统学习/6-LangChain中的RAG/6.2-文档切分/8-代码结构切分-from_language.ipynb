# 代码结构切分：from_language

LangChain 当前没有一个统一的 `CodeTextSplitter` 类。常见方式是使用 `RecursiveCharacterTextSplitter.from_language()`，为不同编程语言选择更合适的分隔符优先级。

```python
from pathlib import Path
from langchain_text_splitters import Language, RecursiveCharacterTextSplitter

path = Path.cwd() / "系统学习" / "6-LangChain中的RAG" / "asset" / "load" / "08-fun_param.py"
source_code = path.read_text(encoding="utf-8")

splitter = RecursiveCharacterTextSplitter.from_language(
    language=Language.PYTHON,
    chunk_size=500,
    chunk_overlap=50,
)
chunks = splitter.create_documents(
    [source_code],
    metadatas=[{"source": str(path), "language": "python"}],
)
```

```python
for chunk in chunks:
    print(chunk.metadata)
    print(chunk.page_content)
    print("-" * 50)
```

`from_language()` 仍然是基于文本分隔符的递归切分，并不真正解析 AST，也不保证函数和类绝不被拆开。需要符号级代码检索时，应使用 tree-sitter、Python `ast` 等解析器先提取函数、类和引用关系，再构造 `Document`。
