# RecursiveJsonSplitter：按 JSON 结构切分

JSON 应优先沿对象和数组结构拆分，而不是先序列化成字符串再任意截断。`RecursiveJsonSplitter` 会尽量保持 JSON 片段合法，并使用 `max_chunk_size` 控制规模。

```python
import json
from pathlib import Path
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveJsonSplitter

CURRENT_DIR = Path.cwd().resolve()
RAG_DIR = next(
    (path for path in (CURRENT_DIR, *CURRENT_DIR.parents) if path.name == "6-LangChain中的RAG"),
    CURRENT_DIR / "系统学习" / "6-LangChain中的RAG",
)
path = RAG_DIR / "asset" / "load" / "03-load.json"
data = json.loads(path.read_text(encoding="utf-8"))

splitter = RecursiveJsonSplitter(max_chunk_size=250)
json_chunks = splitter.split_json(json_data=data)
documents = [
    Document(
        page_content=json.dumps(chunk, ensure_ascii=False),
        metadata={"source": str(path), "chunk_index": index},
    )
    for index, chunk in enumerate(json_chunks)
]
```

```python
for document in documents:
    print(document.metadata)
    print(document.page_content)
```

如果业务已经知道 JSON Schema，通常直接按业务记录构造 `Document` 更可控。`RecursiveJsonSplitter` 更适合结构复杂、层级较深或无法一次放入上下文的通用 JSON。
