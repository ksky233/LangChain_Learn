# 加载 Markdown

Markdown 本质上是纯文本，因此加载阶段直接读取即可。标题层级的识别属于下一阶段的文档切分，不需要在 Loader 中完成。

```python
from pathlib import Path
from langchain_core.documents import Document

ASSET_DIR = Path.cwd() / "系统学习" / "6-LangChain中的RAG" / "asset" / "load"
markdown_path = ASSET_DIR / "06-load.md"

document = Document(
    page_content=markdown_path.read_text(encoding="utf-8"),
    metadata={
        "source": str(markdown_path),
        "format": "markdown",
    },
)

print(document)
```

## 加载与切分要分开

Loader 负责得到完整 Markdown 文本；后续可以使用 `MarkdownHeaderTextSplitter` 按标题切分，并将标题层级写入 metadata。分离职责后，同一个 Loader 可以搭配不同切分策略。
