# 加载 PDF

本例使用轻量的 `pypdf` 提取文本，并按页构造 `Document`。按页保存可以保留页码，便于检索结果引用原文。

```python
from pathlib import Path
from pypdf import PdfReader
from langchain_core.documents import Document

ASSET_DIR = Path.cwd() / "系统学习" / "6-LangChain中的RAG" / "asset" / "load"
pdf_path = ASSET_DIR / "04-load.pdf"

reader = PdfReader(pdf_path)
documents = [
    Document(
        page_content=page.extract_text() or "",
        metadata={
            "source": str(pdf_path),
            "page": page_number,
            "total_pages": len(reader.pages),
        },
    )
    for page_number, page in enumerate(reader.pages, start=1)
]
```

```python
print(f"PDF 页数：{len(documents)}")
print(documents[0].metadata)
print(documents[0].page_content[:500])
```

## 能力边界

`pypdf` 适合包含文本层的普通 PDF。扫描件需要 OCR；复杂表格、双栏排版、公式和图片理解更适合 Docling、Unstructured 或专业 PDF 解析服务。解析质量会直接影响后续 RAG 的召回质量。
