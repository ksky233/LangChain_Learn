# 加载 TXT

TXT 没有复杂结构，直接使用 `pathlib.Path.read_text()` 即可。最重要的是明确文件编码，并把来源路径写入 metadata。

```python
from pathlib import Path
from langchain_core.documents import Document

ASSET_DIR = Path.cwd() / "系统学习" / "6-LangChain中的RAG" / "asset" / "load"

def load_text(path: Path, encoding: str = "utf-8") -> Document:
    return Document(
        page_content=path.read_text(encoding=encoding),
        metadata={"source": str(path), "encoding": encoding},
    )

utf8_doc = load_text(ASSET_DIR / "01-langchain-utf-8.txt")
print(utf8_doc)
```

## 指定非 UTF-8 编码

读取文件时编码错误会导致 `UnicodeDecodeError` 或乱码。编码是输入数据的属性，应由应用明确配置，而不是交给 Embedding 模型处理。

```python
gbk_doc = load_text(
    ASSET_DIR / "01-langchain-gbk.txt",
    encoding="gbk",
)
print(gbk_doc.page_content)
print(gbk_doc.metadata)
```
