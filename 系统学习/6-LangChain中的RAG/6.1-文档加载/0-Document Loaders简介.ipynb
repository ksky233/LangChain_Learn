# 文档加载器 Document Loaders

文档加载的目标，是把 TXT、CSV、JSON、PDF、Word、网页或数据库记录转换成 LangChain 统一的 `Document` 对象。后续的切分、嵌入和检索只处理 `Document`，不需要关心原始数据格式。

本章不使用已经停止维护的 `langchain-community`。简单格式使用 Python 标准库，复杂格式使用轻量专用解析库，最终统一构造 `langchain_core.documents.Document`。

## Document 的基本结构

- `page_content`：真正参与切分、嵌入和检索的文本。
- `metadata`：来源、页码、作者、行号等结构化信息，通常用于过滤、引用和溯源。
- `id`：可选的文档唯一标识。

Loader 的核心职责只有两个：读取原始数据，以及正确构造 `Document`。

```python
from langchain_core.documents import Document

document = Document(
    page_content="LangChain 使用 Document 表示一段文本及其元数据。",
    metadata={"source": "manual", "topic": "RAG"},
    id="doc-1",
)

print(document.page_content)
print(document.metadata)
print(document.id)
```

## 本章实现路线

| 数据格式 | 实现方式 |
| --- | --- |
| TXT、CSV、JSON、Markdown | Python 标准库 + `Document` |
| PDF | `pypdf` + `Document` |
| Word/DOCX | `python-docx` + `Document` |
| HTML | `BeautifulSoup` + `Document` |
| 目录 | `pathlib.rglob()` + 按扩展名分派 |
| 自定义统一接口 | `langchain_core.document_loaders.BaseLoader` |

生产环境遇到复杂版式、表格、图片或 OCR 时，可以进一步选择 `langchain-docling`、`langchain-unstructured` 等独立集成包。
