# 加载 CSV

CSV 通常一行代表一条业务记录。常见做法是每行构造一个 `Document`：用于语义检索的字段放入 `page_content`，ID、作者、行号等字段放入 metadata。

```python
import csv
from pathlib import Path
from langchain_core.documents import Document

CURRENT_DIR = Path.cwd().resolve()
RAG_DIR = next((path for path in (CURRENT_DIR, *CURRENT_DIR.parents) if path.name == "6-LangChain中的RAG"), CURRENT_DIR / "系统学习" / "6-LangChain中的RAG")
ASSET_DIR = RAG_DIR / "asset" / "load"
csv_path = ASSET_DIR / "02-load.csv"

documents = []
with csv_path.open(encoding="utf-8", newline="") as file:
    for row_number, row in enumerate(csv.DictReader(file), start=1):
        documents.append(
            Document(
                page_content=f"标题：{row['title']}\n内容：{row['content']}",
                metadata={
                    "source": str(csv_path),
                    "row": row_number,
                    "id": row["id"],
                    "author": row["author"],
                },
            )
        )
```

```python
print(f"共加载 {len(documents)} 条记录")
for document in documents:
    print(document)
```

CSV 的列不应该全部机械拼进正文。用于精确过滤的字段更适合放 metadata；真正表达语义的标题、描述和正文才适合进入 `page_content`。
