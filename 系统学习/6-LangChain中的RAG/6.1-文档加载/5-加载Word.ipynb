# 加载 Word / DOCX

本例使用 `python-docx` 读取 DOCX。注意 `docx.Document` 表示 Word 文件对象，而 `langchain_core.documents.Document` 表示 RAG 文档，需要使用别名避免重名。

```python
from pathlib import Path
from docx import Document as DocxDocument
from langchain_core.documents import Document

ASSET_DIR = Path.cwd() / "系统学习" / "6-LangChain中的RAG" / "asset" / "load"
docx_path = ASSET_DIR / "05-sgg_chat.docx"
word_file = DocxDocument(docx_path)

paragraphs = [
    paragraph.text.strip()
    for paragraph in word_file.paragraphs
    if paragraph.text.strip()
]

document = Document(
    page_content="\n".join(paragraphs),
    metadata={
        "source": str(docx_path),
        "paragraph_count": len(paragraphs),
    },
)
```

```python
print(document.metadata)
print(document.page_content[:1000])
```

这里只读取普通段落。真实文档还可能包含表格、页眉页脚、图片、批注和文本框；需要保留这些结构时，应使用 Docling、Unstructured 等独立解析方案。
